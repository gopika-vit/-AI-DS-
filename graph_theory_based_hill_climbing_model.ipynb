````python
import random

class GraphProblem:
    def _init_(self):
        # Nodes A–J with some "fitness values"
        self.nodes = {
            "A": 3, "B": 6, "C": 2, "D": 8, "E": 4,
            "F": 10, "G": 1, "H": 7, "I": 5, "J": 9
        }

        # Graph adjacency (neighbors)
        self.edges = {
            "A": ["B", "C"],
            "B": ["A", "D", "E"],
            "C": ["A", "F"],
            "D": ["B", "G"],
            "E": ["B", "H"],
            "F": ["C", "I"],
            "G": ["D", "J"],
            "H": ["E"],
            "I": ["F"],
            "J": ["G"]
        }

    def random_state(self):
        return random.choice(list(self.nodes.keys()))

    def neighbors(self, state):
        return self.edges.get(state, [])

    def evaluate(self, state):
        return self.nodes[state]


def hill_climb(problem, start=None, max_steps=50):
    # Step 1: Initialization
    current = start if start else problem.random_state()
    current_value = problem.evaluate(current)

    path = [current]  # keep track of visited path

    for _ in range(max_steps):
        neighbors = problem.neighbors(current)
        if not neighbors:
            break

        # Pick the best neighbor
        neighbor = max(neighbors, key=problem.evaluate)
        neighbor_value = problem.evaluate(neighbor)

        if neighbor_value <= current_value:
            break  # reached local maximum

        current, current_value = neighbor, neighbor_value
        path.append(current)

    return current, current_value, path


# Run Example
problem = GraphProblem()
solution, value, path = hill_climb(problem, start="A")

print("Best solution found:", solution)
print("Value at solution:", value)
print("Path taken:", " -> ".join(path))
````

````python
# Nodes with values
values = {
    "A": 3, "B": 6, "C": 2, "D": 8, "E": 4,
    "F": 10, "G": 1, "H": 7, "I": 5, "J": 9
}

# Neighbors of each node
neighbors = {
    "A": ["B", "C"],
    "B": ["A", "D", "E"],
    "C": ["A", "F"],
    "D": ["B", "G"],
    "E": ["B", "H"],
    "F": ["C", "I"],
    "G": ["D", "J"],
    "H": ["E"],
    "I": ["F"],
    "J": ["G"]
}

# Hill Climbing Algorithm
def hill_climb(start="A"):
    current = start
    path = [current]

    while True:
        # Find best neighbor
        next_node = max(neighbors[current], key=lambda n: values[n], default=None)

        # If no improvement, stop
        if next_node is None or values[next_node] <= values[current]:
            break

        current = next_node
        path.append(current)

    return current, values[current], path

# Run from A
solution, value, path = hill_climb("A")

print("Best solution:", solution)
print("Value:", value)
print("Path:", " -> ".join(path))
````

````python
# Node values
values = {
    "A": 3, "B": 6, "C": 2, "D": 8, "E": 4,
    "F": 10, "G": 1, "H": 7, "I": 5, "J": 9
}

# Tree neighbors
neighbors = {
    "A": ["B", "C"],
    "B": ["D", "E"],
    "C": ["F"],
    "D": ["G", "H"],
    "E": ["I", "J"],
    "F": [],
    "G": [],
    "H": [],
    "I": [],
    "J": []
}

# Hill Climbing with Goal
def hill_climb_with_goal(start="A", goal="J"):
    current = start
    path = [current]

    while True:
        if current == goal:
            return current, values[current], path, True  # goal reached

        if not neighbors[current]:
            return current, values[current], path, False  # stuck, no neighbors

        # Pick best neighbor
        next_node = max(neighbors[current], key=lambda n: values[n])

        # If no improvement, stuck
        if values[next_node] <= values[current]:
            return current, values[current], path, False

        current = next_node
        path.append(current)

# Run from A to goal J
solution, value, path, reached = hill_climb_with_goal("A", "J")

print("Path taken:", " -> ".join(path))
print("Final Node:", solution)
print("Value:", value)
print("Goal Reached?:", reached)
````

````python
# Node values
values = {
    "A": 3, "B": 6, "C": 2, "D": 8, "E": 4,
    "F": 10, "G": 1, "H": 7, "I": 5, "J": 9
}

# Tree neighbors
neighbors = {
    "A": ["B", "C"],
    "B": ["D", "E"],
    "C": ["F"],
    "D": ["G", "H"],
    "E": ["I", "J"],
    "F": [],
    "G": [],
    "H": [],
    "I": [],
    "J": []
}

# Modified Hill Climb (goal-guided)
def hill_climb_to_goal(start="A", goal="J"):
    current = start
    path = [current]

    while current != goal:
        if not neighbors[current]:
            return current, values[current], path, False  # dead end

        # If goal is in neighbors, move directly
        if goal in neighbors[current]:
            current = goal
            path.append(current)
            return current, values[current], path, True

        # Otherwise, pick the neighbor with the lowest value (to force moving down toward J branch)
        next_node = min(neighbors[current], key=lambda n: values[n])
        current = next_node
        path.append(current)

    return current, values[current], path, True

# Run from A to J
solution, value, path, reached = hill_climb_to_goal("A", "J")

print("Path taken:", " -> ".join(path))
print("Final Node:", solution)
print("Value:", value)
print("Goal Reached?:", reached)
````

````python
from collections import deque

# Node values
values = {
    "A": 3, "B": 6, "C": 2, "D": 8, "E": 4,
    "F": 10, "G": 1, "H": 7, "I": 5, "J": 9
}

# Tree neighbors
neighbors = {
    "A": ["B", "C"],
    "B": ["D", "E"],
    "C": ["F"],
    "D": ["G", "H"],
    "E": ["I", "J"],
    "F": [],
    "G": [],
    "H": [],
    "I": [],
    "J": []
}

# Beam Search
def beam_search(start="A", goal="J", beam_width=2):
    frontier = [(start, [start])]   # (node, path)

    while frontier:
        # Expand all nodes in current frontier
        candidates = []
        for node, path in frontier:
            if node == goal:
                return path, True  # goal found
            for child in neighbors[node]:
                candidates.append((child, path + [child]))

        if not candidates:
            break

        # Keep top-k nodes based on value
        candidates.sort(key=lambda x: values[x[0]], reverse=True)
        frontier = candidates[:beam_width]

    return path, False  # goal not found


# Run Beam Search
path, reached = beam_search(start="A", goal="J", beam_width=2)

print("Path taken:", " -> ".join(path))
print("Goal Reached?:", reached)
````

````python

# Node weights (used as edge costs)
weights = {"A": 3, "B": 6, "C": 2, "D": 8, "E": 4, "F": 10, "G": 1, "H": 7, "I": 5, "J": 9}

# Tree neighbors
neighbors = {
    "A": ["B", "C"], "B": ["D", "E"], "C": ["F"],
    "D": ["G", "H"], "E": ["I", "J"], "F": [], "G": [], "H": [], "I": [], "J": []
}

# Heuristic estimates (to goal J)
h = {"A": 9, "B": 9, "C": 19, "D": 9, "E": 9, "F": 19, "G": 19, "H": 19, "I": 19, "J": 0}

# Beam Search
def beam_search(start, goal, neighbors, weights, h, beam_width=2):
    beam = [(h[start], 0, [start])]  # (f = g + h, g, path)
    while beam:
        new_beam = []
        for _, g, path in beam:
            current = path[-1]
            if current == goal:
                return path, g
            for neighbor in neighbors.get(current, []):
                new_g = g + weights[neighbor]
                new_f = new_g + h[neighbor]
                new_beam.append((new_f, new_g, path + [neighbor]))
        new_beam.sort()
        beam = new_beam[:beam_width]
    return None, None

# Simple text-based tree print
def print_tree(node, neighbors, prefix="", is_last=True):
    print(prefix + ("└── " if is_last else "├── ") + node)
    children = neighbors.get(node, [])
    for i, child in enumerate(children):
        print_tree(child, neighbors, prefix + ("    " if is_last else "│   "), i == len(children) - 1)

# Main execution
start, goal = "A", "J"
print("Tree Structure:")
print_tree(start, neighbors)
print("\nBeam Search Results:")
path, cost = beam_search(start, goal, neighbors, weights, h)
if path:
    print(f"Shortest Path: {path}, Cost: {cost}")
else:
    print("No path found")
```
````

````python
# Tree neighbors
neighbors = {
    "A": ["B", "C"],
    "B": ["D", "E"],
    "C": ["F"],
    "D": ["G", "H"],
    "E": ["I", "J"],
    "F": [], "G": [], "H": [], "I": [], "J": []
}

# Evaluation values for each node (lower = better)
values = {
    "A": 3,
    "B": 6,
    "C": 2,
    "D": 8,
    "E": 4,
    "F": 10,
    "G": 1,
    "H": 7,
    "I": 5,
    "J": 9
}

def beam_search(start="A", goal="J", beam_width=2):
    frontier = [(start, [start])]
    if start == goal:
        return [start], True

    while frontier:
        candidates = []
        for node, path in frontier:
            for child in neighbors[node]:
                new_path = path + [child]
                if child == goal:         # stop if goal found
                    return new_path, True
                candidates.append((child, new_path))

        if not candidates:
            break

        # Sort children by evaluation value (lower is better)
        candidates.sort(key=lambda x: values[x[0]])
        # Keep top beam_width nodes
        frontier = candidates[:beam_width]

    return frontier[0][1], False if frontier else ([start], False)


# Example run
path, reached = beam_search(start="A", goal="J", beam_width=2)
print("Path:", " -> ".join(path))
print("Goal Reached?:", reached)
````
