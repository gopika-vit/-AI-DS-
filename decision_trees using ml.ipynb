```python
# Decision Theory Example
# Actions: Invest in Stock A or Stock B
# Utilities: Profit depending on market condition
actions = {
 "Stock A": {"Good Market": 150, "Bad Market": -50},
 "Stock B": {"Good Market": 30, "Bad Market": 20}
}
# Probabilities of market conditions
prob = {"Good Market": 0.6, "Bad Market": 0.4}
# Compute Expected Utility
expected_utilities = {}
for action, outcomes in actions.items():
 eu = sum(prob[state] * utility for state, utility in outcomes.items())
 expected_utilities[action] = eu
print("Expected Utilities:", expected_utilities)
best_action = max(expected_utilities, key=expected_utilities.get)
print("Best Decision:", best_action)
```

```python
# Bayesian Inference Example
# Disease diagnosis given a symptom
# Prior probability of having disease
P_D = 0.02
P_not_D = 0.98
# Likelihood: Probability of symptom given disease / no disease
P_S_given_D = 0.75
P_S_given_not_D = 0.25
# Evidence: Symptom is observed
P_S = P_S_given_D * P_D + P_S_given_not_D * P_not_D
# Posterior using Bayes' Rule
P_D_given_S = (P_S_given_D * P_D) / P_S
print("Probability of Disease given Symptom:", round(P_D_given_S, 4))
```

```python
# Sequential Decision Problem (MDP)
# Robot moving in a grid with rewards
states = ["A", "B","C", "Goal"]
actions = {"A": {"Right": "B"}, "B": {"Right": "C"}, "C": {"Right": "Goal"}}
rewards = {"A": -2, "B": -1,"C":-1, "Goal": 10}
# Discount factor
gamma = 0.7
# Value iteration
V = {s: 0 for s in states}
for _ in range(10): # 10 iterations
 for s in states:
    if s == "Goal":
       continue
    V[s] = max(rewards[s] + gamma * V[actions[s][a]] for a in actions[s])
print("State Values after iteration:", V)
```

```python
# Game Theory Example - Prisoner's Dilemma
import itertools
players = ["Cooperate", "Defect"]
# Payoff matrix: (Player1, Player2)
payoff = {
 ("Cooperate", "Cooperate"): (8, 8),
 ("Cooperate", "Defect"): (0, 3),
 ("Defect", "Cooperate"): (3, 0),
 ("Defect", "Defect"): (1, 1),
}
print("Payoff Matrix:")
for p1, p2 in itertools.product(players, repeat=2):
 print(f"P1={p1}, P2={p2} -> Payoffs {payoff[(p1, p2)]}")
```

```python
# Medical Diagnosis using Bayesian Inference
diseases = {"Flu": 0.1, "Cold": 0.3, "Allergy": 0.6} # Prior probabilities
symptoms = {
 "Fever": {"Flu": 0.8, "Cold": 0.4, "Allergy": 0.1},
 "Cough": {"Flu": 0.7, "Cold": 0.6, "Allergy": 0.3},
 "Headache": {"Flu": 0.6, "Cold": 0.5, "Allergy": 0.2}
}
# Evidence: Patient has Fever
evidence = "Headache"
P_evidence = sum(symptoms[evidence][d] * diseases[d] for d in diseases)
# Posterior for each disease
posterior = {}
for d in diseases:
 posterior[d] = (symptoms[evidence][d] * diseases[d]) / P_evidence
print("Posterior probabilities given Fever:", posterior)
best_diagnosis = max(posterior, key=posterior.get)
print("Most likely diagnosis:", best_diagnosis)
```
