```python
with open("input.txt", "w") as f:
    f.write("apple banana apple mango banana apple\n")
    f.write("mango apple banana\n")

print("input.txt created")
```

```python
# mapper.py logic

def mapper():
    with open("input.txt", "r") as f:
        for line in f:
            words = line.strip().split()
            for word in words:
                print(f"{word}\t1")

# Run mapper and save output
with open("mapper_output.txt", "w") as out:
    import sys
    original_stdout = sys.stdout
    sys.stdout = out
    mapper()
    sys.stdout = original_stdout

print("Mapper output saved to mapper_output.txt")
```

```python
# Sorting (simulating Hadoop shuffle phase)

with open("mapper_output.txt", "r") as f:
    lines = f.readlines()

lines.sort()

with open("sorted_output.txt", "w") as f:
    f.writelines(lines)

print("Sorted output saved to sorted_output.txt")
```

```python
# reducer.py logic

def reducer():
    current_word = None
    current_count = 0

    with open("sorted_output.txt", "r") as f:
        for line in f:
            word, count = line.strip().split("\t")
            count = int(count)

            if word == current_word:
                current_count += count
            else:
                if current_word:
                    print(f"{current_word}\t{current_count}")
                current_word = word
                current_count = count

    if current_word:
        print(f"{current_word}\t{current_count}")

# Run reducer and save output
with open("final_output.txt", "w") as out:
    import sys
    original_stdout = sys.stdout
    sys.stdout = out
    reducer()
    sys.stdout = original_stdout

print("Final output saved to final_output.txt")
```
